# Parameter Extraction from Natural Language
Extract option parameters from user queries using our LLM-based extraction

```python
import asyncio
from utils.parameter_extraction_utils import extract_from_query
from utils.common import timer
```

## Simple Query

```python
query1 = "Price a call option on AAPL with strike 150, expiring in 30 days"

@timer
async def extract_simple():
    return await extract_from_query(query1)

result1 = await extract_simple()
print(f"Query: {query1}")
print(f"\nExtracted:")
for key, value in result1.items():
    if value is not None:
        print(f"  {key}: {value}")
```

## Complex Query with Context

```python
query2 = "I want to buy a protective put on Tesla, strike around $250, 3 months out, assuming 45% vol"

result2 = await extract_from_query(query2)
print(f"Query: {query2}")
print(f"\nExtracted:")
for key, value in result2.items():
    if value is not None:
        print(f"  {key}: {value}")
```
