# RAG vs Pure LLM Comparison
Compare responses with and without RAG augmentation

```python
import asyncio
from utils.rag_utils import compare_responses, get_rag_sources
```

## Technical Query: Black-Scholes

```python
query = "What are the assumptions of the Black-Scholes model?"

comparison = await compare_responses(query)

print("Query:", query)
print("\n" + "="*50)
print(f"RAG Sources Found: {comparison['sources_found']}")

if comparison['source_metadata']:
    for source in comparison['source_metadata']:
        print(f"  • {source['book']} (p.{source['page']}) - Score: {source['score']:.2f}")

print("\n" + "="*50)
print("RAG-AUGMENTED RESPONSE:")
print(comparison['rag_response'][:500])

print("\n" + "="*50)
print("PURE LLM RESPONSE:")
print(comparison['pure_response'][:500])
```

## Complex Query: Gamma Hedging

```python
query2 = "How do you implement gamma hedging for a portfolio of options?"

comparison2 = await compare_responses(query2)

print("Query:", query2)
print("\n" + "="*50)
print(f"RAG Sources: {comparison2['sources_found']}")
print("\nRAG Response (with textbook sources):")
print(comparison2['rag_response'][:400])
print("\nPure LLM Response (no sources):")
print(comparison2['pure_response'][:400])
```

## Source Quality Analysis

```python
test_queries = [
    "What is delta hedging?",
    "Explain the volatility smile",
    "How do barrier options work?",
    "What is put-call parity?"
]

for q in test_queries:
    sources = get_rag_sources(q)
    print(f"Query: {q}")
    print(f"  Sources: {len(sources)}")
    if sources:
        print(f"  Best match: {sources[0].get('book', 'Unknown')} (score: {sources[0].get('score', 0):.3f})")
    print()
```
